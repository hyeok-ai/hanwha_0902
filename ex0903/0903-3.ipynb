```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def greet(self):
    print("Hello, my name is " + self.name)

  def display_info(self):
    print(f"이름: {self.name}, 나이: {self.age}")

p1 = Person("Emile Jong", 25)
p1.greet()
p1.display_info()

a=p1.display_info
a()
```
