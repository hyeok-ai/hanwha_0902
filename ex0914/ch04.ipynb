```python
from langchain_openai import ChatOpenAI
```

```python
# 쳇모델 객체 생성
llm = ChatOpenAI(
    temperature=0.1,  # 창의성 (0.0 ~ 2.0)
    model_name="gpt-4o-mini",  # 모델명   
)

# 질문
question = "서울이 수도인 국가는 어디인가요?"

# 응답
response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

```python
response.response_metadata["token_usage"]["total_tokens"]
```

### logprob 로그 프로버빌리티 활성화

```python
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini"
).bind(logprobs=True)
```

```python
question = "인도의 수도는 어디인가요?"

response = llm_with_logprob.invoke(question)
```

```python
response.response_metadata["logprobs"]
```

### Stream 실시간 출력 방식

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
```

```python
# flush=False: 버퍼에 저장되었다가 한 번에 출력
# flush=True: 결과가 생성되는대로 바로바로 출력

for token in answer:
    print(token.content, end='', flush=True)
```

```python
answer = llm.stream("성심당 상위 메뉴 10가지와 가격을 알려주세요.")

final_answer = ""
for token in answer:
    final_answer += token.content
```
