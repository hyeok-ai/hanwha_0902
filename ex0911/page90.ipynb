```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    temperature = 0.1,
    model_name = "gpt-4o-mini",
)
```

```python
question = "대한민국 수도는 어디인가요?"

print(f"[답변]: {llm.invoke(question)}")
```

```python
question = "대한민국의 수도는 어디인가요?"

response = llm.invoke(question)

print(type(response))
response

```

```python
print(type(response.content))
response.content
```

```python
print(type(response.response_metadata))
response.response_metadata
```

```python
print(type(response.response_metadata["token_usage"]))
response.response_metadata["token_usage"]
```

```python
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_tokens=2048,
    model_name="gpt-4o-mini",
).bind(logprobs=True)

question = "성심당은 어디에 있나요?"

response = llm_with_logprob.invoke(question)
response.response_metadata
```

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")

for token in answer:
    print(token.content, end="", flush=True)
```

```python
answer = llm.stream("성심당의 대표 메뉴 10가지와 가격을 알려주세요!")

final_answer = ""
for token in answer:
    print(type(token), type(answer)) # <class 'langchain_core.messages.ai.AIMessageChunk'> <class 'generator'>
    final_answer += token.content

print(final_answer)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import pprint

# Gemini 모델 생성
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0
)

# 질의
response = llm.invoke("호두과자 10종류 알려줘.")

print(response.content)
print(response.text())
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# 모델 생성
llm = ChatOpenAI(
    model="gpt-5",
    temperature=0
)

# 시스템 + 유저 메시지
messages = [
    SystemMessage(content="너는 친절한 Python 프로그래밍 강사야."),
    HumanMessage(content="Python에서 리스트와 튜플의 차이를 설명해줘.")
]

# 호출
response = llm.invoke(messages)

# 응답 출력
print(response.content)
```
