```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. NumPy: 가상의 데이터 생성 (랜덤 워크 방식)
# 100일치의 데이터 생성
days = 100
# 정규분포 난수를 누적(cumsum)하여 현실적인 주가 변동 구현
price_fluctuations = np.random.randn(days).cumsum() * 5
base_price = 1000 # 기준 주가
prices = base_price + price_fluctuations

# 2. Pandas: 데이터 프레임 생성 및 데이터 가공
# 2026년 1월 1일부터 100일간의 날짜 데이터 생성
dates = pd.date_range(start="2026-01-01", periods=days, freq='D')

# 날짜와 주가를 엮어 DataFrame 생성
df = pd.DataFrame({
    'Date': dates,
    'Price': prices
})

# Pandas의 강력한 기능인 '이동평균선(Moving Average)' 계산 (7일 기준)
df['7-Day MA'] = df['Price'].rolling(window=7).mean()

# 3. Matplotlib: 데이터 시각화
plt.figure(figsize=(10, 5)) # 그래프 크기 설정

# 원본 주가 그래프 (파란색 얇은 선)
plt.plot(df['Date'], df['Price'], label='Daily Price', color='lightblue', marker='.')

# 7일 이동평균선 그래프 (빨간색 굵은 선)
plt.plot(df['Date'], df['7-Day MA'], label='7-Day Moving Average', color='red', linewidth=2)

# 그래프 꾸미기
plt.title('Mock Stock Price Trend Analysis')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend() # 범례 표시
plt.grid(True, linestyle='--', alpha=0.6) # 배경 격자 추가
plt.xticks(rotation=45) # X축 날짜 글씨가 겹치지 않게 45도 회전
plt.tight_layout() # 레이아웃이 잘리지 않게 조정

# 그래프 출력
plt.show()
```
